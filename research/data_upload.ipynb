```python
import os
import json
import pandas as pd
import pymongo
from dotenv import load_dotenv
import certifi
from insurance.constants.database_variables import DATABASE_NAME, COLLECTION_NAME
from insurance.constants.env_variable import MONGODB_URL_KEY
# Load environment variables
load_dotenv()

# Setup certifi
ca = certifi.where()

# Get MongoDB URI from env
mongo_uri = os.getenv(MONGODB_URL_KEY)

if mongo_uri is None:
    raise Exception("mongo_db_url is not set in .env file")

# Connect to MongoDB
client = pymongo.MongoClient(mongo_uri, tlsCAFile=ca)

# Select DB and collection

database = client[DATABASE_NAME]
collection = database[COLLECTION_NAME]

# Read CSV and insert into MongoDB
df = pd.read_csv("data.csv")
json_data = json.loads(df.to_json(orient="records"))
collection.insert_many(json_data)

print("Data inserted successfully.")

# Close MongoDB connection
client.close()
```
